```python
import pandas as pd

def find_target_genes_quick():
    """
    Quickly check for your target cytokine/chemokine genes
    """
    
    # Your target genes from the list
    target_genes = {
        # Main targets with common mouse/human variants
        'TNF': ['TNF', 'Tnf', 'TNF-α'],
        'IL6': ['IL6', 'Il6', 'IL-6'],
        'IL10': ['IL10', 'Il10', 'IL-10'],
        'IFNG': ['IFNG', 'Ifng', 'IFN-γ', 'Ifnγ'],
        'VEGFA': ['VEGFA', 'Vegfa', 'VEGF-A', 'VEGF', 'Vegf'],
        'CSF2': ['CSF2', 'Csf2', 'GM-CSF'],
        'CSF3': ['CSF3', 'Csf3', 'G-CSF'],
        'CCL2': ['CCL2', 'Ccl2', 'MCP-1'],
        'CXCL10': ['CXCL10', 'Cxcl10', 'IP-10'],
        'CCL21': ['CCL21', 'Ccl21', '6CKine'],
        'IL1B': ['IL1B', 'Il1b', 'IL-1β'],
        'IL1A': ['IL1A', 'Il1a', 'IL-1α'],
        'IL2': ['IL2', 'Il2', 'IL-2'],
        'IL4': ['IL4', 'Il4', 'IL-4'],
        'IL5': ['IL5', 'Il5', 'IL-5'],
        'IL13': ['IL13', 'Il13', 'IL-13'],
        'IL17A': ['IL17A', 'Il17a', 'IL-17A'],
        'TGFB1': ['TGFB1', 'Tgfb1', 'TGF-β1'],
        'CCL5': ['CCL5', 'Ccl5', 'RANTES'],
        'CXCL1': ['CXCL1', 'Cxcl1', 'KC'],
        'CD14': ['CD14', 'Cd14'],
        'TREM1': ['TREM1', 'Trem1'],
        'TNFRSF1A': ['TNFRSF1A', 'Tnfrsf1a', 'TNF-RI'],
        'TNFRSF1B': ['TNFRSF1B', 'Tnfrsf1b', 'TNF-RII'],
    }
    
    print("=== Checking for Target Cytokine/Chemokine Genes ===")
    
    # Read the converted file with correct path
    file_path = '/Users/adityaelayavalli/Downloads/normalized_counts_with_symbols_mygene.csv'
    print(f"Reading: {file_path}")
    
    try:
        df = pd.read_csv(file_path)
        print(f" Dataset loaded: {len(df):,} genes")
        
        # Get all gene symbols, removing NaN values
        all_symbols = df['Gene_Symbol'].dropna().astype(str).tolist()
        symbol_set = set(all_symbols)
        
        print(f"✓ Found {len(all_symbols):,} gene symbols")
        
        # Check for target genes
        found_genes = {}
        not_found = []
        
        print(f"\n=== Target Gene Search Results ===")
        
        for main_name, variants in target_genes.items():
            found = False
            matched_symbol = None
            
            for variant in variants:
                if variant in symbol_set:
                    found = True
                    matched_symbol = variant
                    break
            
            if found:
                gene_info = df[df['Gene_Symbol'] == matched_symbol].iloc[0]
                found_genes[main_name] = {
                    'symbol': matched_symbol,
                    'ensembl_id': gene_info['Genes']
                }
                print(f" {main_name} → {matched_symbol} ({gene_info['Genes']})")
            else:
                not_found.append(main_name)
        
        print(f"\n=== SUMMARY ===")
        print(f"Target genes found: {len(found_genes)}/{len(target_genes)} ({len(found_genes)/len(target_genes)*100:.1f}%)")
        
        if found_genes:
            print(f"\n=== GENES PRESENT IN YOUR DATASET ===")
            for gene, info in found_genes.items():
                print(f"{gene}: {info['symbol']} (Ensembl: {info['ensembl_id']})")
        
        return found_genes
        
    except Exception as e:
        print(f" Error: {e}")
        return None

# Run it
found = find_target_genes_quick()
```
